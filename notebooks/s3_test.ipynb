```python
import boto3
import os 
```

## Upload

```python
s3 = boto3.client("s3")

bucket_name = "ml-demo-amzn"
folder = "../data"

print("Local folder:", os.path.abspath(folder))
print("Files:", os.listdir(folder))

for root, dirs, files in os.walk(folder):

    for filename in files:
        file_path = os.path.join(root, filename)
        

        if os.path.isfile(file_path):

            relative_path = os.path.relpath(file_path, folder)
            s3_key = f"data/{relative_path}".replace("\\", "/")
            
            print(f"Uploading: {file_path}")
            print(f"       to: s3://{bucket_name}/{s3_key}")

            s3.upload_file(
                file_path,
                bucket_name,
                s3_key
            )

print("Upload Successfully.")
```

## Delete

```python
s3 = boto3.client("s3")

bucket_name = "ml-demo-amzn"
s3_key = "data/data.csv"

s3.delete_object(
    Bucket=bucket_name,
    Key=s3_key
)

print("Delete Successfully")
```
